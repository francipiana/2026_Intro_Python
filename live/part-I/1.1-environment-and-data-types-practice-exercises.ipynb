## Exercise 2: Unit conversions

A temperature is recorded as 18.5 °C. Compute it in kelvin and in degrees Fahrenheit. Separately, a discharge of 4.7 m³ s⁻¹ drains a catchment of area 250 km².
````python
temperature_celsius = 18.5
temperature_kelvin = temperature_celsius + 273.15
temperature_fahrenheit = temperature_celsius *9/5 +32
print (f"{temperature_celsius :.2f}")
print (f"{temperature_kelvin :.2f}")
print (f"{temperature_fahrenheit :2f}")
discharge_m3_per_s = 4.7
catchment_area_km2 = 250
discharge_depth_mm = 86.4
specific_discharge_mm = discharge_m3_per_second / catchment_area_km2 * 86.4
print (f"{discharge_m3_per_s :.2f}")
print (f"{catchment_area_km2 :.2f}")
print (f"{discharge_depth_mm :.2f}")
print (f"{specific_discharge_mm :.2f}")
````

## Exercise 4: Parse a record

A logger emits records like `"2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"`. Extract the date string, the station name in title case, the numeric value as a float, and the unit as written, then print them. Then print the station name in lower case as well, the form you would use in a file name.

````python
record = "2024-01-15 , JUNGFRAUJOCH , -2.3 , degC"
pieces = record.split(",")
print(pieces)
date = pieces[0].strip()
station_name = pieces[1].strip().title()
temp_celsius = float (pieces[2].strip())
unit = pieces[3].strip()
print(date)
print(station_name)
print(temp_celsius)
print(unit)
print(station_name.lower())
````

## Exercise 7: Round-trip through a file, with an append

````python
import math
from pathlib import Path
data_path = Path("_files")/"jfj_week.txt"
data_path.parent.mkdir(exist_ok=True)
print(data_path.exists())

with data_path.open("w", encoding="utf-8") as fhandler:
  fhandler.write ("-8.4\n")
  fhandler.write ("-8.9\n")
  fhandler.write ("-9.2\n")
  fhandler.write ("-9.0\n")
  fhandler.write ("-9.5\n")

with data_path.open("a", encoding="utf-8") as fhandler:
  fhandler.write ("-7.8\n")
  fhandler.write ("-6.1\n")

with data_path.open ("r", encoding="utf-8") as fhandler:
  all_lines = fhandler.readlines()
print(len(all_lines))
print(all_lines)

day_1_temp_celsius = all_lines [0]
print(day_1_temp_celsius)
day_2_temp_celsius = all_lines[1]
print(day_2_temp_celsius)
day_3_temp_celsius = all_lines[2]
print(day_3_temp_celsius)
day_4_temp_celsius = all_lines[3]
print(day_4_temp_celsius)
day_5_temp_celsius = all_lines[4]
print(day_5_temp_celsius)
day_6_temp_celsius = all_lines[5]
print(day_6_temp_celsius)
day_7_temp_celsius = all_lines[6]
print(day_7_temp_celsius)

day_1_float_temp_celsius = float(day_1_temp_celsius)
day_2_float_temp_celsius = float(day_2_temp_celsius)
day_3_float_temp_celsius = float(day_3_temp_celsius)
day_4_float_temp_celsius = float(day_4_temp_celsius)
day_5_float_temp_celsius = float(day_5_temp_celsius)
day_6_float_temp_celsius = float(day_6_temp_celsius)
day_7_float_temp_celsius = float(day_7_temp_celsius)

mean_daily_temp_celsius = (day_1_float_temp_celsius + day_2_float_temp_celsius + day_3_float_temp_celsius + day_4_float_temp_celsius + day_5_float_temp_celsius + day_6_float_temp_celsius + day_7_float_temp_celsius) / (len(all_lines))
print(f"{mean_daily_temp_celsius :.2f}")



````
